### Age Category

```python
def cat_age(age):
    if age <= 10:
        return 'Child'
    elif age <= 35:
        return 'Youth'
    elif age <= 55:
        return 'Adult'
    else:
        return 'Senior'
data['Age_Group'] = data['Age'].apply(cat_age)
```

```python
data.head()
```

### Bar Chart Distribution of the Age Group

```python
Age_Counts = data['Age_Group'].value_counts().sort_index()

plt.bar(Age_Counts.index, Age_Counts.values, color = 'skyblue')
plt.title('Age Bracket Distribution')
plt.xlabel('Age Braket')
plt.ylabel('No. of Passengers')
plt.show()
```

### Flight Status per Continent

```python
status_by_continent = data.groupby(['Airport_Continent','Flight_Status']).size().unstack()
status_by_continent
```

```python
status_by_continent.plot(kind='bar',stacked=True, figsize=(8,5),color=['red','yellow','green'])
plt.title('Flight Status by ontinent')
plt.xlabel('Continent')
plt.ylabel('No. of Flight')
plt.show()
```

### Flight Delay Ratio per Continent

```python
delay_rate = (
    data.groupby('Airport_Continent')['Flight_Status']
    .apply(lambda x: (x == 'Delayed').mean() * 100)
    .sort_values(ascending = False)
)
```

```python
delay_rate.plot(kind='bar',color='teal')
plt.title('Delay Rate (%) by continent')
plt.xlabel('Continent')
plt.ylabel('Percentage of Delayed Flights')
plt.show()
```

### Total Flights per Day, Delayed Flight per Day

```python
#Convert Datetime format
data['Departure_Date'] = pd.to_datetime(data['Departure_Date'])
```

```python
data.head(1)
```

```python
data['Departure_Day'] = data['Departure_Date'].dt.day_name()
```

```python
data.head(2)
```

```python
#Total Flights per Day
data['Departure_Day'].value_counts()
```

```python
#Chart of Total Fligt per Day
data['Departure_Day'].value_counts().reindex(
    ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
).plot(kind='bar',color='blue')
plt.title('No. of Flights per Departure Day')
plt.xlabel('Day of the Wee')
plt.ylabel('No. of Flghts')
plt.show()
```

```python
#Delayed Flights per Day
data[data['Flight_Status'] == 'Delayed']['Departure_Day'].value_counts().plot(kind='bar', color='red')
plt.show()
```

```python
#Days wth high chances o Flight Cancellation
cancel_rate_by_day = (
    data.groupby('Departure_Day')['Flight_Status']
    .apply(lambda x: (x == 'Cancelled').mean()*100)
    .reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
    )
cancel_rate_by_day
```

```python
delay_rate_by_day = (
    data[data['Flight_Status'].notna()]
    .groupby('Departure_Day')['Flight_Status']
    .apply(lambda x: (x == 'Delayed').sum() / len(x) * 100)
    .reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
    )
delay_rate_by_day
```

### Time Series Analysis

```python
#Daily Count of Delayed Flights
daily_delays = (
    data[data['Flight_Status'] == 'Delayed']
    .groupby(data['Departure_Date'].dt.date)
    .size()
)
daily_delays
```

```python
#Daily Rate of Delay (%)
daily_delay_rate = (
    data.groupby(data['Departure_Date'].dt.date)['Flight_Status']
        .apply(lambda x:(x=='Delayed').mean()*100)
)
daily_delay_rate
```

```python
#Daily Delay Rate Line Chart
daily_delay_rate.plot(kind='line',figsize=(10,5))
plt.title('Daily Delay Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Delay Rate (%)')
plt.grid(True)
plt.show()
```
